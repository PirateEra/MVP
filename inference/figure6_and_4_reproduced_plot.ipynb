# Figure 4 reproduced


```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.ticker as ticker
```

```python
# flops dl19 220m model MVP
# 197983445625792

# flops dl19 220m model ListT5
# 1113282018800640

# flops amount of ListT5 relative to MVP
rel_flops = round((1113282018800640/197983445625792), 1)

# results
models = ['ListT5', 'MVP', 'ListT5(reproduced)', 'MVP(reproduced)']
flops = [5.5, 1.0, rel_flops, 1.0]
ndcg = [70.0, 71.8, 70.0, 71.7]

fig, ax1 = plt.subplots(figsize=(16,10))

# add labels to bars
for i in range(len(models)):
  ax1.text(models[i], flops[i]+0.075, f'x{flops[i]}', ha='center')
# set range of y axis of the barplot
ax1.set_ylim(0, 6)
# bar plot
fig1 = ax1.bar(models, flops, color='lightcoral', width=0.4, label='FLOPs')
ax1.set_ylabel('FLOPs', weight='bold')

# set up second axis for the second line plot
ax2 = ax1.twinx()
# add labels to the data points
for i in range(len(models)):
  ax2.text(models[i], ndcg[i]+0.075, f'x{ndcg[i]}', ha='center')
# set range of y axis of the line plot
ax2.set_ylim(68, 72.25)
# set ticks steps to be 1
ax2.yaxis.set_major_locator(ticker.MultipleLocator(1))
# line plot
fig2 = ax2.plot(models, ndcg, marker='o', label='TREC-DL Avg')
ax2.set_ylabel('nDCG@10', weight='bold')

# add legend to plot
figs = [fig1] + fig2
labels = [figure.get_label() for figure in figs]
ax1.legend(figs, labels, loc=0)
plt.show()
```

# Figure 6 reproduced


```python
# results where the 4-view model is re-trained
views = np.array([1, 2, 3, 4, 5, 6])
ndcg = np.array([51.2, 50.8, 51.3, 50.9, 50.4, 50.5])
original = np.array([50.9, 51.3, 51.2, 51.4, 51.2, 50.6])

plt.figure(figsize=(12, 7))

for i in range(len(views)):
  plt.text(views[i], ndcg[i]+0.035, ndcg[i], ha='center')

for i in range(len(views)):
  plt.text(views[i], original[i]+0.035, original[i], ha='center')

plt.plot(views, ndcg, linestyle='dashed', marker='o', label='reproduced')
plt.plot(views, original, marker='o', label='original')
plt.yticks(np.arange(50.0, 52.5, 0.5))
plt.xlabel('Number of View Tokens', weight='bold')
plt.ylabel('nDCG@10', weight='bold')
plt.title("BEIR Avg.", weight='bold')
plt.legend()
plt.show()
```
